```python
import pandas as pd
from data_profiling import ProfileReport
import os

path_to_save_profiles = rf"../ignore_files/data/profiles/"

list_directory = os.listdir("../ignore_files/data/seperated/")

# Scans directory ../ignore_files/data/seperated/ for CSVs, puts them in an array as dataframes
dataframes = [pd.read_csv(os.path.join(root, file)) for root, _, files in os.walk("../ignore_files/data/seperated/") for file in files if file.endswith(".csv")]
file_names = [file for root, _, files in os.walk("../ignore_files/data/seperated/") for file in files if file.endswith(".csv")]

# Schema
type_schema = {
    "x": "numeric",
    "y": "numeric",
    "z": "numeric",
    "Creation_Time": "numeric",
    "Arrival_Time": "numeric",
    "Sensor_Time": "numeric"
}


for i, df in enumerate(dataframes):
    file_name = file_names[i]
    # Creates a profile report for each dataframe
    profile = ProfileReport(df, title=f"Profile Report for DataFrame {file_name}", explorative=True, type_schema=type_schema)
    
    # Saves the profile report as an HTML file
    profile.to_file(os.path.join(path_to_save_profiles, f"{file_name}.html"))
```

```python
import pandas as pd
import numpy as np

# Expected sample rates (Hz)
sample_rates = {
    "nexus4": 200,
    "s3": 150,
    "s3mini": 100,
    "samsungold": 50,
    "lgwatch": 200,
    "gear": 100
}

df = pd.read_csv('../ignore_files/data/full/Phones_accelerometer.csv', index_col=0)
df_ = pd.read_csv('../ignore_files/data/full/Watch_accelerometer.csv', index_col=0)

df = pd.concat([df, df_], axis=0, ignore_index=True)

# 1. Convert timestamps to floored datetime objects (1-second precision)
# Arrival_Time is in milliseconds; Creation_Time is in nanoseconds
df['Arrival_Sec'] = pd.to_datetime(df['Arrival_Time'], unit='ms').dt.floor('s')
df['Creation_Sec'] = pd.to_datetime(df['Creation_Time'], unit='ns').dt.floor('s')

group_keys = ['User', 'Model', 'Device']

# 2. Count samples per second for Arrival_Time
arr_counts = (
    df.groupby(group_keys + ['Arrival_Sec'])
    .size()
    .reset_index(name='arrival_hz')
)

# 3. Count samples per second for Creation_Time
cre_counts = (
    df.groupby(group_keys + ['Creation_Sec'])
    .size()
    .reset_index(name='creation_hz')
)

# 4. Optional: Trim incomplete first/last seconds per device session
# Edge seconds often contain partial sampling windows that artificially lower Hz
def trim_edge_seconds(counts_df, sec_col):
    return counts_df.groupby(group_keys, group_keys=False).apply(
        lambda g: g[(g[sec_col] > g[sec_col].min()) & (g[sec_col] < g[sec_col].max())]
    )

arr_counts_trimmed = trim_edge_seconds(arr_counts, 'Arrival_Sec')
cre_counts_trimmed = trim_edge_seconds(cre_counts, 'Creation_Sec')

# 5. Compute summary metrics per (User, Model, Device) group
arr_stats = arr_counts_trimmed.groupby(group_keys)['arrival_hz'].agg(
    arr_mean='mean',
    arr_std='std',
    arr_min='min',
    arr_max='max'
).reset_index()

cre_stats = cre_counts_trimmed.groupby(group_keys)['creation_hz'].agg(
    cre_mean='mean',
    cre_std='std',
    cre_min='min',
    cre_max='max'
).reset_index()

# 6. Merge stats and compare with expected sampling rates
report = pd.merge(arr_stats, cre_stats, on=group_keys)
report['expected_hz'] = report['Model'].map(sample_rates)

# Calculate percentage drift from expected Hz
report['arr_drift_%'] = ((report['arr_mean'] - report['expected_hz']) / report['expected_hz']) * 100
report['cre_drift_%'] = ((report['cre_mean'] - report['expected_hz']) / report['expected_hz']) * 100

# Flag integrity status (True if within 5% tolerance)
TOLERANCE_PCT = 5.0
report['arrival_passed'] = report['arr_drift_%'].abs() <= TOLERANCE_PCT
report['creation_passed'] = report['cre_drift_%'].abs() <= TOLERANCE_PCT

# 7. Save outputs for audit and verification
arr_counts.to_csv("sample_counts_per_second_arrival.csv", index=False)
cre_counts.to_csv("sample_counts_per_second_creation.csv", index=False)
report.to_csv("sample_rate_integrity_report.csv", index=False)

print(report.to_string())
```

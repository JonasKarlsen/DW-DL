```python
import pandas as pd

df_watch_acc = pd.read_csv("../ignore_files/data/full/Watch_accelerometer.csv", index_col=0)
df_phone_acc = pd.read_csv("../ignore_files/data/full/Phones_accelerometer.csv", index_col=0)
df_watch_gyro = pd.read_csv("../ignore_files/data/full/Watch_gyroscope.csv", index_col=0)
df_phone_gyro = pd.read_csv("../ignore_files/data/full/Phones_gyroscope.csv", index_col=0)

df_acc = pd.concat([df_watch_acc, df_phone_acc], ignore_index=True)
df_gyro = pd.concat([df_watch_gyro, df_phone_gyro], ignore_index=True)

# Split into an array of dataframes based on unique combinations of User, Model, and Device
group_cols = ["User", "Model", "Device"]

acc_groups = [
    group.reset_index(drop=True)
    for _, group in df_acc.groupby(group_cols, sort=False)
]
gyro_groups = [
    group.reset_index(drop=True)
    for _, group in df_gyro.groupby(group_cols, sort=False)
]

# Save each dataframe to a CSV file with a unique name based on the User, Model, and Device
for i, group in enumerate(acc_groups):
    user = group["User"].iloc[0]
    model = group["Model"].iloc[0]
    device = group["Device"].iloc[0]
    filename = f"acc_group_{i}_{user}_{model}_{device}.csv"
    group.to_csv(filename, index=False)

for i, group in enumerate(gyro_groups):
    user = group["User"].iloc[0]
    model = group["Model"].iloc[0]
    device = group["Device"].iloc[0]
    filename = f"gyro_group_{i}_{user}_{model}_{device}.csv"
    group.to_csv(filename, index=False)
```

```python
import pandas as pd


df_watch_acc = pd.read_csv("../ignore_files/data/full/Watch_accelerometer.csv", index_col=0)
df_watch_gyro = pd.read_csv("../ignore_files/data/full/Watch_gyroscope.csv", index_col=0)

df_phone_acc = pd.read_csv("../ignore_files/data/full/Phones_accelerometer.csv", index_col=0)
df_phone_gyro = pd.read_csv("../ignore_files/data/full/Phones_gyroscope.csv", index_col=0)

full_df = pd.concat(
    [df_watch_acc, df_watch_gyro, df_phone_acc, df_phone_gyro],
    ignore_index=True,
)

unique_combos = full_df[['Model', 'Device']].drop_duplicates()
unique_combo__ = full_df[['Device']].drop_duplicates()
unique_combo___ = full_df[['Model']].drop_duplicates()

for i in unique_combos:
    print(i)

print(unique_combos)
print(unique_combo__)
print(unique_combo___)
```

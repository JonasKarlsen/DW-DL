```python
import pandas as pd

df = pd.read_csv("../ignore_files/data/full/Phones_accelerometer.csv", index_col=0)

df.info()

# Drop columns where gt is null
df = df.dropna(subset=['gt'])
```

```python
df.head()
```

```python
devices = df['Device'].unique()

users = df['User'].unique()

model = df['Model'].unique()

activities = df['gt'].unique()

for activity in activities:
    print(f"Activity: {activity}, Count: {len(df[df['gt'] == activity])}")

for device in devices:
    print(f"Device: {device}, Count: {len(df[df['Device'] == device])}")

for model in model:
    print("ALLAH")
    print(f"Model: {model}, Count: {len(df[df['Model'] == model])}")
    print("ALLAH")

for user in users:
    print(f"User: {user}, Count: {len(df[df['User'] == user])}")

print(f"Devices: {devices}"
      )

print(f"Users: {users}"
      )

print(f"Models: {model}"
      )

# Count each device grouped by user
device_user_counts = df.groupby(['User', 'Device']).size().reset_index(name='Count')

# Print the counts as CSV so they can be copied into Excel
print(device_user_counts.to_csv(index=False), end="")

# # Count each activity grouped by user
# activity_user_counts = df.groupby(['User', 'gt']).size().reset_index(name='Count')

# # Print the counts as CSV so they can be copied into Excel
# print(activity_user_counts.to_csv(index=False), end="")

# # Print the count of each row where gt column is null / NaN
# null_gt_count = df['gt'].isnull().sum()
# print(f"Count of rows where gt is null: {null_gt_count}")

# # Print the count of each row where Device column is null / NaN
# null_device_count = df['Device'].isnull().sum()
# print(f"Count of rows where Device is null: {null_device_count}")

# # Print the count of each row where Model column is null / NaN
# null_model_count = df['Model'].isnull().sum()
# print(f"Count of rows where Model is null: {null_model_count}")

# # Print the count of each row where User column is null / NaN
# null_user_count = df['User'].isnull().sum()
# print(f"Count of rows where User is null: {null_user_count}")

# Make a dataframe where each row is a device and each column is a user
device_user_matrix = (
    df.groupby(['Device', 'User'])
      .size()
      .unstack(fill_value=0)
)

print()

print(device_user_matrix.to_csv())


```
